```python
#동물병원

import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/corazzon/boostcourse-ds-510/master/data/medical_201909.csv.zip", low_memory=False)

# 동물병원 수 찾기
ani_count= df[df['상권업종소분류명'] == '동물병원'].groupby('시도명').size().sort_values(ascending=False)

# 동물병원 수 시각화
import matplotlib.pyplot as plt
plt.rc('font', family='NanumGothic')

ani_count.plot(kind='bar', figsize=(10, 6), color='skyblue')
plt.title('시도별 동물병원 수')
plt.ylabel('병원 수')
plt.xticks(rotation=30)
plt.show()

print(ani_count)
```

```python
#2-1.진짜로 상관관계가 있을까?

import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/corazzon/boostcourse-ds-510/master/data/medical_201909.csv.zip", low_memory=False)

# 시도별 병원 및 약국 수 집계
hp_count = df[df['상권업종중분류명'].isin(['병원', '약국'])].groupby('시도명').size()

# 시도별 동물병원 수 집계
ani_count= df[df['상권업종소분류명'] == '동물병원'].groupby('시도명').size().sort_values(ascending=False)

# 병원 및 약국 수와 동물병원 수 데이터를 하나의 df로 결합
correlation_df = pd.DataFrame(
    {
    '병원_약국_수': hp_count,
    '동물병원_수': ani_count
    }
    )

#결측치 개수 확인
null_count=correlation_df.isnull().sum()
print(null_count) #결측치 0으로 채워서 둘 다 0이 뜹니다! 유의 부탁드려요

# 결측치를 0으로 채우고 계산
correlation_df = correlation_df.fillna(0)
correlation = correlation_df.corr()

print(correlation)
```

```python
# 2-2 산포도로 확인해보기

import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/corazzon/boostcourse-ds-510/master/data/medical_201909.csv.zip", low_memory=False)

# 종류 확인
print(df.head(3))

# 시도별 병원 및 약국 수 집계
hospital_drug_count = df[df['상권업종중분류명'].isin(['병원', '약국'])].groupby('시도명').size()

# 시도별 동물병원 수 집계
animal_hospital_count = df[df['상권업종소분류명'] == '동물병원'].groupby('시도명').size()

# 병원 및 약국 수와 동물병원 수 데이터를 하나의 데이터프레임으로 결합
correlation_df = pd.DataFrame(
    {
    '병원_약국_수': hospital_drug_count,
    '동물병원_수': animal_hospital_count
    }
    )

# 결측치를 0으로 채우기 (동물병원이 없는 경우 등)
correlation_df = correlation_df.fillna(0)

# 산포도 시각화
import matplotlib.pyplot as plt
plt.rc('font', family='NanumGothic')

plt.figure(figsize=(10, 6))
plt.scatter(correlation_df['병원_약국_수'], correlation_df['동물병원_수'], color='skyblue', edgecolor='black')

# 그래프에 라벨 추가
for i in range(correlation_df.shape[0]):
    plt.text(correlation_df['병원_약국_수'][i], correlation_df['동물병원_수'][i], correlation_df.index[i], fontsize=9)

plt.title('Scattered map of hospital and pharmacy  no. corrrelation')
plt.xlabel('병원/약국 수')
plt.ylabel('동물병원 수')
plt.grid(True)
plt.show()
```

#📌Q3. 강남지역에는 다른 지역에 비해 피부과나 성형외과가 많아 보입니다. 실제로 해당 지역에 피부과나 성형외과가 다른 지역에 비해 전체 병원 수 중에서 어느 정도의 비율을 차지하고 있는지 알아보겠습니다.


*   서울에 소재한 병원 중 상권업종소분류명에 "피부" 나 "성형"이 들어간 분류명을 찾아 구해주세요!
*   피부과 성형외과 수 / 전체병원 수 로 비율을 구해주세요!

*   비율이 높은 순서대로 정렬되게 구해주세요!
*   수업에서 사용한 같은 CSV 파일을 사용하며 다음의 결과가 나오도록 구합니다.
소숫점 두 번째짜리까지 출력하는 방법은 pandas round 로 검색해서 사용법을
알아보세요!



```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/corazzon/boostcourse-ds-510/master/data/medical_201909.csv.zip", low_memory=False)

# 서울특별시에서 상권업종소분류명이 피부 또는 성형인 것을 찾아서 시군구명을 loc로 찾아서 x에 저장
x = df.loc[(df['시도명'] == '서울특별시') & (df['상권업종소분류명'].str.contains('피부|성형')), '시군구명']

# 서울특별시에서 상권업종중분류명이 병원인 것을 찾아서 시군구명을 loc로 찾아서 y에 저장
y = df.loc[(df['시도명'] == '서울특별시') & (df['상권업종중분류명'].str.contains('병원')), '시군구명']

# x의 value_counts()를 y의 value_counts()로 나누고, 이를 내림차순으로 정렬하고, 소수점 둘째자리까지 반올림하여 출력
print((x.value_counts()/y.value_counts()).sort_values(ascending=False).round(2))
```

#📌Q4. 평균 기대수명이 점점 길어지면서 실버 의료 산업도 주목받고 있습니다. 여러분은 실버 의료 산업과 관련된 스타트업에 취업했습니다. 지도를 시각화하여 '노인/치매병원'이 주로 어디에 위치하고 있는지를 찾아보고자 합니다.


*   folium 을 통해 지도에 전국의 '노인/치매병원'을 표시해 주세요!



```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/corazzon/boostcourse-ds-510/master/data/medical_201909.csv.zip", low_memory=False)

# 노인/치매병원에 해당하는 데이터프레임 만들기
df_old = df[df["상권업종소분류명"] == "노인/치매병원"].copy()
```

```python
 import folium
 map = folium.Map(location=[df_old["위도"].mean(),df_old["경도"].mean()],zoom_start=8)

for i in df_old.index:
  name = df_old.loc[i,"상호명"]
  address = df_old.loc[i,"도로명주소"]
  popup = f"{name}-{address}"
  location = [df_old.loc[i,"위도"],df_old.loc[i,"경도"]]
  folium.Marker(
      location = location,
      popup = popup,
  ).add_to(map)
map
```
